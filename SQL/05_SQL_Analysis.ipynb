<font color="green" size="5"><b>SQL Business Analysis</b></font>

In this step, SQL queries are used to analyze the final analytical dataset stored in the database.

The objective is to answer key business questions using SQL and support data-driven decision making.

The following analyses are performed:

- Top 10 customers by revenue
- Revenue by category
- Monthly growth rate
- Cancellation rate per country
- Payment success rate

The dataset used for these queries is the **Final_Dataset** table stored in the SQLite database.

```python
import sqlite3
import pandas as pd

df = pd.read_csv("../Data/Final_Dataset.csv")

conn = sqlite3.connect("E-Commerce.db")

df.to_sql("Final_Dataset", conn, if_exists="replace", index=False)
```

```python
query = "SELECT * FROM Final_Dataset LIMIT 5"

pd.read_sql(query, conn)
```

```python
# Top 10 customers by revenue (Which customers generate the most revenue)

query = """
SELECT 
    customer_id,
    SUM(revenue) AS total_revenue
FROM Final_Dataset
WHERE valid_order = 1
GROUP BY customer_id
ORDER BY total_revenue DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# Revenue by category (Which product categories are the most profitable)

query = """
SELECT 
    category,
    SUM(revenue) AS total_revenue
FROM Final_Dataset
WHERE valid_order = 1
GROUP BY category
ORDER BY total_revenue DESC
"""

pd.read_sql(query, conn)
```

```python
# Monthly Revenue Growth Rate (How is revenue growing month over month)

query = """
SELECT 
    strftime('%Y-%m', order_date) AS month,
    SUM(revenue) AS monthly_revenue
FROM Final_Dataset
WHERE valid_order = 1
GROUP BY month
ORDER BY month
"""

pd.read_sql(query, conn)
```

```python
# Cancellation Rate per Country (Which countries have the highest cancellation rates)

query = """
SELECT 
    country,
    COUNT(CASE WHEN status = 'Cancelled' THEN 1 END) * 100.0 / COUNT(*) AS cancellation_rate
FROM Final_Dataset
GROUP BY country
ORDER BY cancellation_rate DESC
"""

pd.read_sql(query, conn)
```

```python
# Payment Success Rate (Which payment methods have the highest success rates)

query = """
SELECT 
    payment_method,
    COUNT(CASE WHEN payment_status = 'Success' THEN 1 END) * 100.0 / COUNT(*) AS success_rate
FROM Final_Dataset
GROUP BY payment_method
ORDER BY success_rate DESC
"""

pd.read_sql(query, conn)
```
