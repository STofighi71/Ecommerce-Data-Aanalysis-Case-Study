<font color="green" size="5"><b>KPI Engineering</b></font>


In this section, key business metrics (KPIs) are calculated to evaluate the overall performance of the e-commerce business.

These metrics help measure sales performance, customer behavior, and operational efficiency.

The KPIs calculated in this analysis include:

• Total Revenue  
• Net Revenue (excluding returns and cancellations)  
• Average Order Value (AOV)  
• Customer Lifetime Value (basic estimation)  
• Churn Proxy (inactive customers)  
• Conversion Rate (successful payments)

These metrics will later be used in the business dashboard to monitor company performance and support data-driven decision making.

<font color="green" size="4"><b>Importing necessary libraries and Loading the datasets</b></font>

```python
import pandas as pd

df = pd.read_csv("../Data/Final_Dataset.csv")
```

<font color="green" size="4"><b>Total Revenue (VALID ONLY)(Delivered + Payment Success)</b></font>

```python
total_revenue = df[df["valid_order"] == True]["revenue"].sum()

total_revenue.round(2)
```

<font color="green" size="4"><b>Net Revenue (Exclude Returns)</b></font>

```python
# Calculate net revenue by filtering for delivered orders with successful payments
# This metric reflects the actual realized sales.

net_revenue = df[
    (df["status"] == "Delivered") &
    (df["payment_status"] == "Success")
]["revenue"].sum()

net_revenue.round(2)
```

<font color="green" size="4"><b>Average Order Value</b></font>

```python
# Calculate average order value (AOV) by dividing total revenue from valid orders by the number of unique valid orders
# AOV helps evaluate customer purchasing behavior and sales efficiency.

avg_order_value = df[df["valid_order"] == True]["revenue"].sum() / df[df["valid_order"] == True]["order_id"].nunique()

avg_order_value.round(2)
```

<font color="green" size="4"><b>Customer Lifetime Value (Basic)</b></font>

```python
# Calculate customer lifetime value (CLV) by grouping revenue by customer_id and taking the mean
# A basic estimation of the total revenue generated by each customer across all orders.

clv = df.groupby("customer_id")["revenue"].sum().mean()

clv.round(2)
```

<font color="green" size="4"><b>Churn Proxy (inactive customers if no orders in last 90 days)</b></font>

```python
# Calculate churn proxy by identifying customers whose last order date is older than a specified cutoff (e.g., 90 days)
# This metric helps identify customers who may have stopped engaging with the platform.
# Customers who have not placed orders within a defined time period are considered inactive.

df["order_date"] = pd.to_datetime(df["order_date"])

last_order = df.groupby("customer_id")["order_date"].max()

cutoff = df["order_date"].max() - pd.Timedelta(days=90)

churn_customers = last_order[last_order < cutoff].count()

churn_customers
```

<font color="green" size="4"><b>Conversion Rate (Payment Success)</b></font>

```python
# Calculate conversion rate by determining the percentage of successful payments out of total orders
# This metric indicates the effectiveness of the sales process in converting orders into successful transactions.

conversion_rate = (df["payment_status"] == "Success").mean() * 100

conversion_rate.round(2)
```
