<font color="green" size="5"><b>Data Modeling</b></font>

In this step, the different datasets are joined to create a final analytical table.
This table will be used for analysis, KPI calculations, and dashboard creation.

The following tables are joined:
- customers
- orders
- order_items
- products
- payments

<font color="green" size="4"><b>Importing necessary libraries</b></font>

```python
import pandas as pd
import numpy as np
```

<font color="green" size="4"><b>Loading the datasets</b></font>

```python
customers = pd.read_csv("../Data/customers_clean.csv")
products = pd.read_csv("../Data/products.csv")
orders = pd.read_csv("../Data/orders_clean.csv")
order_items = pd.read_csv("../Data/order_items.csv")
payments = pd.read_csv("../Data/payments.csv")
```

<font color="green" size="4"><b>Validating Relationships (orders ↔ customers)
(order_items ↔ products)
</b></font>


```python
# Validate that all customer_ids in orders exist in customers (check orders → customers relationship)
# This ensures that every order is associated with a valid customer.
orders["customer_id"].isin(customers["customer_id"]).all()
```

```python
# Validate that all product_ids in order_items exist in products (check order_items → products relationship)
# This ensures that every order item references a valid product.

order_items["product_id"].isin(products["product_id"]).all()
```

<font color="green" size="4"><b>Joining tables to create a final analytical table</b></font>


```python
# Join Orders + Customers

df = orders.merge(customers, on="customer_id", how="left")
```

```python
# Join Order Items

df = df.merge(order_items, on="order_id", how="left")
```

```python
# Join Products

df = df.merge(products, on="product_id", how="left")
```

```python
# Join Payments

df = df.merge(payments, on="order_id", how="left")
```

<font color="green" size="4"><b>Creating Revenue Column</b></font>


```python
df["revenue"] = df["quantity"] * df["price"]
```

<font color="green" size="4"><b>Creating Valid Order</b></font>


```python
df["valid_order"] = (
    (df["status"] == "Delivered") &
    (df["payment_status"] == "Success")
)
```
