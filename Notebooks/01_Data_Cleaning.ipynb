<font color="green" size="4"><b>Filling the missing values in the country column with "Unknown" to preserve customer records</b></font>

```python
customers["country"] = customers["country"].fillna("Unknown")
customers.isnull().sum()
```

<font color="green" size="4"><b>Removing Duplicates from Orders Dataset</b></font>

```python
orders = orders.drop_duplicates()
orders.duplicated().sum()
```
